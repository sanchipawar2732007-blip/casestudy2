```python
# Social Network using Adjacency Matrix

users = ["A", "B", "C", "D", "E"]

n = len(users)

# Create an adjacency matrix
matrix = [[0 for j in range(n)] for i in range(n)]


# Function to add friendship
def add_friendship(user1, user2):
    if user1 in users and user2 in users:
        i = users.index(user1)
        j = users.index(user2)

        matrix[i][j] = 1
        matrix[j][i] = 1

        print(user1, "and", user2, "are now friends.")
    else:
        print("User not found!")


# Function to display adjacency matrix
def display_matrix():
    print("\nAdjacency Matrix:")
    print("  ", users)

    for i in range(n):
        print(users[i], matrix[i])


# Function to display friends of a user
def display_friends(user):
    if user not in users:
        print("User not found!")
        return

    i = users.index(user)
    print("\nFriends of", user, ":")

    found = False

    for j in range(n):
        if matrix[i][j] == 1:
            print(users[j])
            found = True

    if not found:
        print("No friends found.")


# Function to check friendship
def check_friendship(user1, user2):
    if user1 not in users or user2 not in users:
        print("User not found!")
        return

    i = users.index(user1)
    j = users.index(user2)

    if matrix[i][j] == 1:
        print(user1, "and", user2, "are friends.")
    else:
        print(user1, "and", user2, "are not friends.")


# Function to count friends
def count_friends(user):
    if user not in users:
        print("User not found!")
        return

    i = users.index(user)
    count = sum(matrix[i])

    print("Total friends of", user, "=", count)


# Add friendship connections
add_friendship("A", "B")
add_friendship("A", "C")
add_friendship("B", "C")
add_friendship("B", "D")
add_friendship("C", "E")
add_friendship("D", "E")

# Display all connections
display_matrix()

# Display friends
display_friends("B")

# Check friendship
check_friendship("A", "D")
check_friendship("C", "E")

# Count total friends
count_friends("B")
count_friends("E")
```
